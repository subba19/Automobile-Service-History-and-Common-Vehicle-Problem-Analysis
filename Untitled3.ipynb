Machine Learning – Data preprocessing

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

X = df[
    ["CITY", "STATE", "SERVICE HISTORY", "VEHICAL COMPANY"]
]

y = df["COMMON PROBLEM"]

# Combine classes with fewer than 5 records
rare_classes = y.value_counts()
rare_classes = rare_classes[rare_classes < 5].index

y = y.where(~y.isin(rare_classes), "Other/Rare Problems")

print("Features selected:")
print(X.columns.tolist())

print("\nTarget variable:", "COMMON PROBLEM")
```

Train and test split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Records:", len(X_train))
print("Testing Records:", len(X_test))
```

Build Machine Learning model

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("city", OneHotEncoder(handle_unknown="ignore"), ["CITY"]),
        ("state", OneHotEncoder(handle_unknown="ignore"), ["STATE"]),
        ("service", TfidfVectorizer(token_pattern=r"(?u)\b\w+\b"),
         "SERVICE HISTORY"),
        ("company", OneHotEncoder(handle_unknown="ignore"),
         ["VEHICAL COMPANY"])
    ]
)

model = Pipeline([
    ("preprocessing", preprocessor),
    ("classifier", LogisticRegression(
        max_iter=3000,
        class_weight="balanced"
    ))
])

model.fit(X_train, y_train)

print("Machine Learning Model Trained Successfully!")
```

Prediction

```python
y_pred = model.predict(X_test)

results = pd.DataFrame({
    "Actual Problem": y_test.values,
    "Predicted Problem": y_pred
})

display(results.head(20))
```

Model evaluation

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

accuracy = accuracy_score(y_test, y_pred)

precision = precision_score(
    y_test, y_pred,
    average="weighted",
    zero_division=0
)

recall = recall_score(
    y_test, y_pred,
    average="weighted",
    zero_division=0
)

f1 = f1_score(
    y_test, y_pred,
    average="weighted",
    zero_division=0
)

print("MODEL PERFORMANCE")

print("Accuracy:", round(accuracy * 100, 2), "%")
print("Precision:", round(precision * 100, 2), "%")
print("Recall:", round(recall * 100, 2), "%")
print("F1 Score:", round(f1 * 100, 2), "%")

print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))
```

Confusion matrix

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(
    y_test,
    y_pred,
    labels=model.classes_
)

plt.figure(figsize=(12, 8))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=model.classes_,
    yticklabels=model.classes_
)

plt.title("Confusion Matrix")
plt.xlabel("Predicted Values")
plt.ylabel("Actual Values")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

Performance comparison graph

```python
metrics = {
    "Accuracy": accuracy * 100,
    "Precision": precision * 100,
    "Recall": recall * 100,
    "F1 Score": f1 * 100
}

plt.figure(figsize=(8, 5))

plt.bar(metrics.keys(), metrics.values())

plt.title("Machine Learning Performance")
plt.ylabel("Percentage")
plt.ylim(0, 100)

plt.show()
```
